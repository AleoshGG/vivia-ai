# 01 — Extracción de Datos y EDA

**Objetivo:** Explorar los datasets de Properati MX (renta y venta), mapear columnas al
esquema `Draft`, y extraer los parámetros reales (medias, desviaciones, rangos por tipo de
propiedad) que calibrarán el dataset artificial de entrenamiento.

**Datasets:**
- `../datasets/properati-MX-2016-11-01-properties-rent.csv` (~40 MB)
- `../datasets/properati-MX-2016-11-01-properties-sell.csv` (~185 MB)

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import warnings

warnings.filterwarnings('ignore')
pd.set_option('display.max_columns', 50)
pd.set_option('display.float_format', '{:,.2f}'.format)
sns.set_theme(style='whitegrid')

RENT_PATH  = '../datasets/properati-MX-2016-11-01-properties-rent.csv'
SELL_PATH  = '../datasets/properati-MX-2016-11-01-properties-sell.csv'
```

## 1. Carga de datos

```python
df_rent = pd.read_csv(RENT_PATH, low_memory=False)
df_sell = pd.read_csv(SELL_PATH, low_memory=False)

print(f'Renta  → {df_rent.shape[0]:,} filas, {df_rent.shape[1]} columnas')
print(f'Venta  → {df_sell.shape[0]:,} filas, {df_sell.shape[1]} columnas')
print('\nColumnas disponibles:')
print(df_rent.columns.tolist())
```

```python
# Unimos renta y venta; solo conservamos las columnas relevantes para el mapeo
COLS_KEEP = [
    'operation', 'property_type', 'place_name',
    'price', 'price_aprox_local_currency', 'currency',
    'price_per_m2', 'price_usd_per_m2',
    'surface_total_in_m2', 'surface_covered_in_m2',
    'rooms',
]

cols_rent = [c for c in COLS_KEEP if c in df_rent.columns]
cols_sell = [c for c in COLS_KEEP if c in df_sell.columns]

df = pd.concat([
    df_rent[cols_rent],
    df_sell[cols_sell],
], ignore_index=True)

print(f'Dataset combinado: {df.shape[0]:,} filas')
df.head(3)
```

## 2. Análisis de nulos

```python
null_pct = (df.isnull().sum() / len(df) * 100).sort_values(ascending=False)
null_df = pd.DataFrame({'% nulos': null_pct, 'count': df.isnull().sum()})
print(null_df[null_df['count'] > 0].to_string())
```

## 3. Mapeo de columnas Properati → esquema Draft

| Campo `Draft` | Columna Properati | Transformación |
|---|---|---|
| `areaM2` | `surface_total_in_m2` | float, dropna |
| `listedPrice` | `price_aprox_local_currency` | float MXN |
| `pricePerM2` | derivado | `listedPrice / areaM2` |
| `bedrooms` | `rooms` | int, imputar mediana |
| `bathrooms` | — | ratio estimado por tipo |
| `parkingSpaces` | — | imputar por tipo |
| `propertyType.name` | `property_type` | normalizar texto |
| `address.postalCode` | `place_name` | proxy de zona |
| `constructionYear` | — | distribución por tipo |

```python
# Seleccionar columna de área (preferir surface_total, fallback surface_covered)
if 'surface_total_in_m2' in df.columns:
    df['areaM2'] = pd.to_numeric(df['surface_total_in_m2'], errors='coerce')
    # Completar nulos con surface_covered
    if 'surface_covered_in_m2' in df.columns:
        mask = df['areaM2'].isna()
        df.loc[mask, 'areaM2'] = pd.to_numeric(df.loc[mask, 'surface_covered_in_m2'], errors='coerce')

# Precio en MXN
if 'price_aprox_local_currency' in df.columns:
    df['listedPrice'] = pd.to_numeric(df['price_aprox_local_currency'], errors='coerce')
else:
    df['listedPrice'] = pd.to_numeric(df['price'], errors='coerce')

# pricePerM2 derivado
if 'price_per_m2' in df.columns:
    df['pricePerM2'] = pd.to_numeric(df['price_per_m2'], errors='coerce')
mask_ppm2 = df['pricePerM2'].isna() & df['areaM2'].notna() & (df['areaM2'] > 0)
df.loc[mask_ppm2, 'pricePerM2'] = df.loc[mask_ppm2, 'listedPrice'] / df.loc[mask_ppm2, 'areaM2']

# Habitaciones
df['bedrooms'] = pd.to_numeric(df.get('rooms'), errors='coerce')

# Tipo de propiedad normalizado
if 'property_type' in df.columns:
    df['propertyType'] = df['property_type'].str.strip().str.title()
else:
    df['propertyType'] = 'Unknown'

# Zona proxy
df['zona'] = df.get('place_name', pd.Series('Desconocido', index=df.index)).fillna('Desconocido')

print('Tipos de propiedad únicos:')
print(df['propertyType'].value_counts())
```

```python
# Filtrar filas con features mínimas necesarias
df_clean = df.dropna(subset=['areaM2', 'listedPrice', 'pricePerM2']).copy()

# Quitar outliers extremos antes del EDA (precios y áreas absurdas)
df_clean = df_clean[
    (df_clean['areaM2'] > 10) & (df_clean['areaM2'] < 10_000) &
    (df_clean['listedPrice'] > 1_000) & (df_clean['listedPrice'] < 100_000_000) &
    (df_clean['pricePerM2'] > 100) & (df_clean['pricePerM2'] < 500_000)
]

print(f'Dataset limpio: {df_clean.shape[0]:,} filas')
df_clean[['areaM2', 'listedPrice', 'pricePerM2', 'bedrooms', 'propertyType']].describe()
```

## 4. Distribuciones por tipo de propiedad

```python
TOP_TYPES = df_clean['propertyType'].value_counts().head(5).index.tolist()
df_top = df_clean[df_clean['propertyType'].isin(TOP_TYPES)]

fig, axes = plt.subplots(1, 3, figsize=(18, 5))
features = ['areaM2', 'listedPrice', 'pricePerM2']
labels   = ['Área (m²)', 'Precio listado (MXN)', 'Precio/m² (MXN)']

for ax, feat, label in zip(axes, features, labels):
    for pt in TOP_TYPES:
        subset = df_top[df_top['propertyType'] == pt][feat]
        ax.hist(subset, bins=60, alpha=0.5, label=pt)
    ax.set_xlabel(label)
    ax.set_ylabel('Frecuencia')
    ax.legend(fontsize=7)
    ax.set_title(f'Distribución de {label}')

plt.tight_layout()
plt.show()
```

## 5. Correlaciones

```python
corr_cols = ['areaM2', 'listedPrice', 'pricePerM2', 'bedrooms']
corr_matrix = df_clean[corr_cols].corr()

plt.figure(figsize=(7, 5))
sns.heatmap(corr_matrix, annot=True, fmt='.2f', cmap='coolwarm', vmin=-1, vmax=1)
plt.title('Correlaciones entre features numéricas')
plt.tight_layout()
plt.show()
```

## 6. Extracción de parámetros reales por tipo de propiedad

Estos parámetros calibrarán el dataset artificial en el notebook 02.

```python
def iqr_bounds(series, factor=1.5):
    """Retorna (lower, upper) usando IQR para filtrar outliers."""
    q1, q3 = series.quantile(0.25), series.quantile(0.75)
    iqr = q3 - q1
    return q1 - factor * iqr, q3 + factor * iqr

params = {}
for pt in TOP_TYPES:
    sub = df_clean[df_clean['propertyType'] == pt]
    if len(sub) < 50:
        continue

    # Rangos IQR para cada feature
    area_lo, area_hi   = iqr_bounds(sub['areaM2'])
    price_lo, price_hi = iqr_bounds(sub['listedPrice'])
    ppm2_lo, ppm2_hi   = iqr_bounds(sub['pricePerM2'])

    params[pt] = {
        'n_samples'           : len(sub),
        'area_mean'           : sub['areaM2'].mean(),
        'area_std'            : sub['areaM2'].std(),
        'area_min'            : max(10, area_lo),
        'area_max'            : area_hi,
        'price_mean'          : sub['listedPrice'].mean(),
        'price_std'           : sub['listedPrice'].std(),
        'price_min'           : max(1_000, price_lo),
        'price_max'           : price_hi,
        'price_per_m2_mean'   : sub['pricePerM2'].mean(),
        'price_per_m2_std'    : sub['pricePerM2'].std(),
        'price_per_m2_min'    : max(100, ppm2_lo),
        'price_per_m2_max'    : ppm2_hi,
        'bedrooms_median'     : sub['bedrooms'].median() if sub['bedrooms'].notna().any() else 2.0,
        'bedrooms_std'        : sub['bedrooms'].std()    if sub['bedrooms'].notna().any() else 1.0,
    }

import json
print(json.dumps(
    {k: {f: round(v, 2) if isinstance(v, float) else v for f, v in vals.items()}
     for k, vals in params.items()},
    indent=2, ensure_ascii=False
))
```

```python
import json, pathlib

out_path = pathlib.Path('../datasets/eda_params.json')
out_path.write_text(json.dumps(params, indent=2, ensure_ascii=False, default=float))
print(f'Parámetros guardados en: {out_path.resolve()}')
```

## 7. Resumen del mapeo

| Campo `Draft` | Fuente | Estado |
|---|---|---|
| `areaM2` | `surface_total_in_m2` | ✅ disponible |
| `listedPrice` | `price_aprox_local_currency` | ✅ disponible |
| `pricePerM2` | derivado | ✅ disponible |
| `bedrooms` | `rooms` | ✅ disponible (% nulos) |
| `bathrooms` | — | ⚠️ ausente → estimar por tipo |
| `parkingSpaces` | — | ⚠️ ausente → estimar por tipo |
| `constructionYear` | — | ⚠️ ausente → distribución por tipo |
| `propertyType.name` | `property_type` | ✅ disponible |
| `address.postalCode` | `place_name` | ✅ disponible (proxy) |
| `condominium` | — | ⚠️ ausente → derivar de tipo |
| `amenityIds` | — | ⚠️ ausente → generar random |

Los campos ausentes se imputarán con distribuciones razonables en el notebook 02.
